#**Probabilidad Condicional**

##Ejercicio 1
Se lanza un dado equilibrado. Calcula la probabilidad de obtener un 2, dado que salió un número par.

```python
P_A = 1/6  # Probabilidad de obtener un 2
P_B_A = 1  # Probabilidad de par si salió un 2
P_B = 3/6  # Probabilidad de obtener un número par

P_A_B = (P_B_A * P_A) / P_B

print("La probabilidad de obtener un 2 dado que salió un número par es:")
print(P_A_B)
print("Porcentaje:", P_A_B * 100, "%")
```

##Ejercicio 2
Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

```python
total = 0
favorables = 0

for d1 in range(1, 7):
    for d2 in range(1, 7):
        for d3 in range(1, 7):

            if d1 != d2 and d1 != d3 and d2 != d3:
                total += 1

                if d1 + d2 + d3 == 9:
                    favorables += 1

probabilidad = favorables / total

print("Casos posibles:", total)
print("Casos favorables:", favorables)
print("Probabilidad:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")
```

#**Probabilidad Total**

#Ejercicio 1
El 70% de los clientes ven un anuncio y, de ellos, compra el 30%. El 30% no ve el anuncio y, de ellos, compra el 10%. Calcula la probabilidad total de que un cliente compre.

```python
# Definimos las probabilidades
ve_anuncio = 0.70
compra_ve = 0.30
no_ve_anuncio = 0.30
compra_no_ve = 0.10

# Aplicamos la fórmula de probabilidad total
probabilidad_compra = (
    compra_ve * ve_anuncio +
    compra_no_ve * no_ve_anuncio
)

# Mostramos el resultado
print(f"Probabilidad de compra: {probabilidad_compra:.2%}")
```

##Ejercicio 2
Se sabe que:

El 40% de los días son nublados

De esos, el 75% llueve

El 60% de los días no son nublados

De esos, el 10% llueve

¿Cuál es la probabilidad total de que llueva?

```python
# Definimos las probabilidades
dia_nublado = 0.40
lluvia_nublado = 0.75
dia_no_nublado = 0.60
lluvia_no_nublado = 0.10

# Aplicamos la fórmula de probabilidad total
probabilidad_lluvia = (
    lluvia_nublado * dia_nublado +
    lluvia_no_nublado * dia_no_nublado
)

# Mostramos el resultado
print(f"Probabilidad de lluvia: {probabilidad_lluvia:.2%}")
```

##Ejercicio 3
Hay dos cajas de audífonos. La primera tiene 2 que sirven y 6 que fallan. La segunda tiene 5 que sirven y 3 que fallan. Se elige una caja al azar y después un audífono. ¿Cuál es la probabilidad de que sirva?

```python
# Datos de la primera caja
sirven_caja1 = 2
fallan_caja1 = 6

# Datos de la segunda caja
sirven_caja2 = 5
fallan_caja2 = 3

# Probabilidad de elegir cada caja
prob_caja1 = 0.50
prob_caja2 = 0.50

# Calculamos la probabilidad de que sirva un audífono
prob_sirve_caja1 = sirven_caja1 / (sirven_caja1 + fallan_caja1)
prob_sirve_caja2 = sirven_caja2 / (sirven_caja2 + fallan_caja2)

# Aplicamos la fórmula de probabilidad total
probabilidad_sirve = (
    prob_sirve_caja1 * prob_caja1 +
    prob_sirve_caja2 * prob_caja2
)

# Mostramos los resultados
print(f"Probabilidad de servir en la caja 1: {prob_sirve_caja1:.2%}")
print(f"Probabilidad de servir en la caja 2: {prob_sirve_caja2:.2%}")
print(f"Probabilidad total de que sirva: {probabilidad_sirve:.2%}")
```

##**Teorema de Bayes**

##Ejercicio 1
El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta», ¿cuál es la probabilidad de que sea spam?

```python
# Definimos las probabilidades del problema
P_spam = 0.20
P_oferta_spam = 0.80
P_oferta_deseado = 0.10
P_deseado = 1 - P_spam

# Calculamos la probabilidad total de encontrar la palabra "oferta"
P_oferta = (P_oferta_spam * P_spam) + (P_oferta_deseado * P_deseado)

# Aplicamos el teorema de Bayes
P_spam_oferta = (P_oferta_spam * P_spam) / P_oferta

# Mostramos los resultados
print(f"Probabilidad de encontrar 'oferta': {P_oferta:.2%}")
print(f"Probabilidad de que sea spam: {P_spam_oferta:.2%}")
```

##Ejercicio 2
En una tienda online, el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.

Si se registró una venta, ¿qué navegador es más probable que haya usado el cliente?

```python

# Navegador más probable en una compra

# Probabilidad de utilizar cada navegador
P_chrome = 0.40
P_safari = 0.35
P_otros = 0.25

# Tasas de conversión de cada navegador
P_venta_chrome = 0.03
P_venta_safari = 0.04
P_venta_otros = 0.02

# Calculamos la probabilidad total de que exista una venta
P_venta = (
    P_venta_chrome * P_chrome
    + P_venta_safari * P_safari
    + P_venta_otros * P_otros
)

# Aplicamos el teorema de Bayes para cada navegador
P_chrome_venta = (P_venta_chrome * P_chrome) / P_venta
P_safari_venta = (P_venta_safari * P_safari) / P_venta
P_otros_venta = (P_venta_otros * P_otros) / P_venta

# Guardamos los resultados en un diccionario
resultados = {
    "Chrome": P_chrome_venta,
    "Safari": P_safari_venta,
    "Otros navegadores": P_otros_venta
}

# Mostramos las probabilidades
print(f"Probabilidad total de venta: {P_venta:.2%}\n")

for navegador, probabilidad in resultados.items():
    print(f"{navegador}: {probabilidad:.2%}")

# Identificamos el navegador más probable
navegador_mas_probable = max(resultados, key=resultados.get)

print("\nNavegador más probable:", navegador_mas_probable)
print(f"Probabilidad: {resultados[navegador_mas_probable]:.2%}")
```
